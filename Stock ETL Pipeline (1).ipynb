```python
for symbol in all_data:
    for date in all_data[symbol]["Time Series (Daily)"]:
        stock = all_data[symbol]["Time Series (Daily)"][date]
    
        records.append({
            "symbol": symbol,
            "date": date,
            "open": stock["1. open"], "high": stock["2. high"], "low": stock["3. low"], "close": stock["4. close"]
    })
```
